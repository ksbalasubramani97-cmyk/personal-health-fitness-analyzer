```python
# ==========================================
# PERSONAL HEALTH & FITNESS ANALYZER
# ==========================================

print("=" * 50)
print("       PERSONAL HEALTH & FITNESS ANALYZER")
print("=" * 50)

# -----------------------------
# USER INPUT
# -----------------------------

gender = input("Enter your Gender (Male/Female): ").strip().lower()

height = float(input("Enter your Height in cm: "))
weight = float(input("Enter your Weight in kg: "))

# -----------------------------
# BMI CALCULATION
# -----------------------------

height_m = height / 100

bmi = weight / (height_m * height_m)

print("\n" + "=" * 50)
print("              YOUR HEALTH REPORT")
print("=" * 50)

print("Gender       :", gender.capitalize())
print("Height       :", height, "cm")
print("Weight       :", weight, "kg")
print("BMI          :", round(bmi, 2))

# -----------------------------
# UNDERWEIGHT
# -----------------------------

if bmi < 18.5:

    category = "Underweight"

    print("\nBMI Category :", category)

    print("\nPossible Health Risks:")
    print("- Nutritional deficiencies")
    print("- Weak immune system")
    print("- Low bone strength")
    print("- Muscle weakness")
    print("- Anemia risk")
    print("- Fatigue and low energy")

    print("\nFoods to Include:")
    print("- Eggs")
    print("- Milk and curd")
    print("- Nuts and seeds")
    print("- Banana")
    print("- Rice")
    print("- Whole grains")
    print("- Dal and beans")
    print("- Paneer")
    print("- Peanut butter")
    print("- Vegetables and fruits")

    print("\nFoods to Limit:")
    print("- Excess junk food")
    print("- Excess sugary drinks")
    print("- Skipping meals")

    print("\nHome Workout:")
    print("- Bodyweight squats")
    print("- Wall push-ups")
    print("- Lunges")
    print("- Light strength exercises")
    print("- Stretching")

    print("\nGoal:")
    print("Focus on balanced nutrition and gradual healthy weight gain.")


# -----------------------------
# NORMAL WEIGHT
# -----------------------------

elif bmi >= 18.5 and bmi < 25:

    category = "Normal Weight"

    print("\nBMI Category :", category)

    print("\nPossible Health Risks:")
    print("- Weight-related health risks are generally lower")
    print("- However, BMI does not guarantee absence of disease")
    print("- Other factors such as diet, activity and family history matter")

    print("\nFoods to Include:")
    print("- Vegetables")
    print("- Fruits")
    print("- Whole grains")
    print("- Eggs")
    print("- Fish or lean protein")
    print("- Dal and beans")
    print("- Nuts and seeds")
    print("- Milk or curd")
    print("- Adequate water")

    print("\nFoods to Limit:")
    print("- Excess fried food")
    print("- Excess sugar")
    print("- Excess processed food")
    print("- Sugary soft drinks")

    print("\nHome Workout:")
    print("- Walking")
    print("- Jogging")
    print("- Squats")
    print("- Push-ups")
    print("- Plank")
    print("- Stretching")

    print("\nGoal:")
    print("Maintain a balanced diet and regular physical activity.")


# -----------------------------
# OVERWEIGHT
# -----------------------------

elif bmi >= 25 and bmi < 30:

    category = "Overweight"

    print("\nBMI Category :", category)

    print("\nPossible Health Risks:")
    print("- Higher risk of type 2 diabetes")
    print("- High blood pressure")
    print("- Heart disease")
    print("- High cholesterol")
    print("- Joint problems")
    print("- Sleep-related breathing problems")

    print("\nFoods to Include:")
    print("- Vegetables")
    print("- Fruits in appropriate portions")
    print("- Whole grains")
    print("- Dal")
    print("- Beans")
    print("- Eggs")
    print("- Fish")
    print("- Lean protein")
    print("- Nuts in small portions")
    print("- Plenty of water")

    print("\nFoods to Limit:")
    print("- Deep-fried foods")
    print("- Fast food")
    print("- Sugary drinks")
    print("- Excess sweets")
    print("- Chips")
    print("- Highly processed foods")
    print("- Excess refined carbohydrates")

    print("\nHome Workout:")
    print("- Brisk walking")
    print("- Marching in place")
    print("- Bodyweight squats")
    print("- Wall push-ups")
    print("- Step-ups")
    print("- Light jogging if comfortable")
    print("- Stretching")

    print("\nGoal:")
    print("Focus on gradual weight management through healthy eating and regular activity.")


# -----------------------------
# OBESE
# -----------------------------

else:

    category = "Obesity"

    print("\nBMI Category :", category)

    print("\nPossible Health Risks:")
    print("- Higher risk of type 2 diabetes")
    print("- High blood pressure")
    print("- Heart disease")
    print("- Stroke")
    print("- High cholesterol")
    print("- Sleep apnea")
    print("- Joint problems")
    print("- Fatty liver disease")

    print("\nFoods to Include:")
    print("- Plenty of vegetables")
    print("- Whole fruits")
    print("- Whole grains")
    print("- Dal and beans")
    print("- Eggs")
    print("- Fish")
    print("- Lean protein")
    print("- Low-fat dairy")
    print("- Nuts in controlled portions")
    print("- Water")

    print("\nFoods to Limit:")
    print("- Deep-fried foods")
    print("- Fast food")
    print("- Sugary drinks")
    print("- Sweets")
    print("- Cakes and pastries")
    print("- Chips")
    print("- Highly processed foods")
    print("- Excess alcohol")

    print("\nHome Workout:")
    print("- Walking")
    print("- Marching in place")
    print("- Chair exercises")
    print("- Wall push-ups")
    print("- Gentle squats")
    print("- Low-impact exercises")
    print("- Stretching")

    print("\nGoal:")
    print("Focus on gradual, sustainable weight management.")

# -----------------------------
# GENDER INFORMATION
# -----------------------------

print("\n" + "=" * 50)
print("              GENDER NOTE")
print("=" * 50)

if gender == "male":

    print("Male:")
    print("- Maintain regular physical activity.")
    print("- Include sufficient protein and vegetables.")
    print("- Get adequate sleep and hydration.")

elif gender == "female":

    print("Female:")
    print("- Maintain regular physical activity.")
    print("- Include iron-rich and protein-rich foods.")
    print("- Include vegetables, fruits and whole grains.")
    print("- Get adequate sleep and hydration.")

else:

    print("Gender not recognized.")
    print("General recommendations will still apply.")

# -----------------------------
# FINAL MESSAGE
# -----------------------------

print("\n" + "=" * 50)
print("                  FINAL NOTE")
print("=" * 50)

print("BMI is only a screening measurement.")
print("It cannot diagnose a disease by itself.")
print("Health risks depend on many factors.")
print("For medical concerns, consult a qualified healthcare professional.")

print("\nThank you for using Personal Health & Fitness Analyzer!")
print("=" * 50)
```
